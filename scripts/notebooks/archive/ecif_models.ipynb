Feature importance using boruta

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from boruta import BorutaPy
df = pd.read_csv("Trial/ECIF_with_activity_train.csv") 
df = df.dropna(subset=["pIC"]).copy()

feature_names = df.drop(columns=["Complex_ID", "pIC"]).columns.tolist()
X = np.nan_to_num(df.drop(columns=["Complex_ID", "pIC"]).values.astype(np.float64))
y = df["pIC"].values.astype(np.float64)

rf = RandomForestRegressor(n_estimators=200, max_depth=7, n_jobs=-1, random_state=42)

boruta = BorutaPy(
    estimator=rf,
    n_estimators="auto",
    max_iter= 100,
    random_state=42,
    verbose=2,
)
boruta.fit(X, y)

# ranking_: 1 = most important, higher = less important
out = pd.DataFrame({
    "feature": feature_names,
    "boruta_rank": boruta.ranking_,
})
out = out.sort_values("boruta_rank")  # best first
out.to_csv("boruta_feature_ranks.csv", index=False)

print(out.head(30))
print("Saved boruta_feature_ranks.csv")
```

```python
# features with rank 1 = confirmed
confirmed_mask = boruta.ranking_ == 1

rf.fit(X, y)
imp = rf.feature_importances_

out = pd.DataFrame({
    "feature": feature_names,
    "boruta_rank": boruta.ranking_,
    "rf_importance": imp,
})
# sort: confirmed first, then by RF importance
out = out.sort_values(["boruta_rank", "rf_importance"], ascending=[True, False])
print(out.head(30))
```

```python
import matplotlib.pyplot as plt

top = out.head(29).iloc[::-1]  # already sorted; reverse for barh

fig, ax = plt.subplots(figsize=(7, 8))
ax.barh(top["feature"], top["rf_importance"])
ax.set_xlabel("RF importance")
ax.set_ylabel("Feature")
ax.set_title("Top important features by Boruta")
fig.tight_layout()
fig.savefig("Fig_boruta_top29.png", dpi=300, bbox_inches="tight")
plt.show()
```

Hyperparameter Tuning of XGBoost model

```python
from sklearn.model_selection import GridSearchCV
import xgboost as xgb

# Step 1: Define model (GPU version)
xgb_model = xgb.XGBRegressor(tree_method = "hist", device = "cuda",random_state=42)

# Step 2: Define hyperparameters to tune
n_estimators = list(range(1,151))   
max_depth = list(range(1,7))
learning_rate = [0.01, 0.05, 0.1]
gamma = [0, 0.5, 1]
min_child_weight = list(range(1,5))

# Step 3: Create the parameter grid for GridSearchCV
tuning_parameters = dict(
    n_estimators=n_estimators,
    max_depth=max_depth,
    learning_rate=learning_rate,
    gamma=gamma,
    min_child_weight=min_child_weight
)

# Step 4: Setup GridSearchCV
grid_search = GridSearchCV(estimator=xgb_model,param_grid=tuning_parameters,cv=5,scoring="neg_mean_absolute_error", n_jobs=-1,verbose=2)

# Step 5: Fit on training data
grid_search.fit(X_train, y_train)

# Step 6: Results
print("✅ Best parameters:", grid_search.best_params_)
print("✅ Best CV score:", grid_search.best_score_)
```

```python
# Train–external PCA (ECIF space) — uses FULL training cohort, not 80% split
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

train = pd.read_csv("/media/user/282895b1-543c-4689-9ced-b38fcef5e554/YASHI_ECIF/Trial/ECIF_with_activity_train.csv")
ext   = pd.read_csv("/media/user/282895b1-543c-4689-9ced-b38fcef5e554/YASHI_ECIF/Trial/ECIF_descriptors_test_clean.csv")

feat = [c for c in train.columns if c not in ("Complex_ID", "pIC")]
Xtr = train[feat].fillna(0).values.astype(float)
Xex = ext[feat].fillna(0).values.astype(float)

sc = StandardScaler().fit(Xtr)          # fit on FULL train cohort
Ztr = sc.transform(Xtr)
Zex = sc.transform(Xex)

pca = PCA(n_components=2, random_state=42).fit(Ztr)
PCtr = pca.transform(Ztr)
PCex = pca.transform(Zex)

plt.figure(figsize=(6, 5))
plt.scatter(PCex[:, 0], PCex[:, 1], s=8, alpha=0.35, label=f"External (n={len(PCex)})")
plt.scatter(PCtr[:, 0], PCtr[:, 1], s=8, alpha=0.45, label=f"Train cohort (n={len(PCtr)})")
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)")
plt.title("ECIF descriptor space: train vs external")
plt.legend()
plt.tight_layout()
plt.show()
```
